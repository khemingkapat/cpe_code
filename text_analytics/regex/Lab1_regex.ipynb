# 3. Regular expression tokenizer

## 3.1 Basic regex tokenizer

````python
s = ("Good muffins cost $3.88\nin New York.  Please buy me\n"
      "two of them.\n\nThanks.")
s2 = ("Alas, it has not rained today. When, do you think, "
      "will it rain again?")
s3 = ("<p>Although this is <b>not</b> the case here, we must "
      "not relax our vigilance!</p>")
````

````python
s2
````

````python
nltk.regexp_tokenize(s2, r'[,\.\?!"]\s*', gaps=True)
````

What happened after applying the above RegEx tokenizer to the input string?

Ans: it is like splitting with either `, ` `. ` `? ` `!` or any of the symbols in the set following with white space

````python
s3
````

````python
nltk.regexp_tokenize(s3, r'</?.>', gaps=False)
````

What happened after applying the above RegEx tokenizer to the input string?

Ans: to match anything with that pattern. within `<>` with either zero or one `/` the `gaps=False` means that we ignore others and match that only. like we not splitting with what we specify, we want that as our result and use the negative as splitter

````python
nltk.regexp_tokenize(s3, r'</?.>', gaps=True)
````

What happened after applying the above RegEx tokenizer to the input string?

Ans: as the negative one of the previous one. we splitting with the bracket thing there.

# 4. Harry Potter Things

````python
def head(collection,n=5):
    return collection[:n]
    
    
````

````python
raw_lines = raw.split("\n")
````

````python
lines = [item for item in raw_lines if item]
````

````python
head(lines)
````

````python
book_count = 0
for idx,line in enumerate(lines):
    if line == "Harry Potter":
        print(f"book no. {book_count} >>",end="")
        print("at idx : ",idx,">>",line,end = " ")
        print(lines[idx+1])
        book_count += 1
````

````python
books = dict()
current_book = ""
for idx,line in enumerate(lines):
    if line == "Harry Potter":
        current_book = line + " " + lines[idx+1]
        books[current_book] = {"content" : [line],
                               "start" : idx,
                               "end" : idx+1
                              }
    else:
        books[current_book]["content"].append(line)
        books[current_book]["end"] += 1
````

````python
for book_name,book_item in books.items():
    print(book_name,book_item["start"],book_item["end"])
````

````python
test_string = "“something something” something said harry"
test_string_2 = "harry something something “something something”"


pattern = r'Harry[^“”\n]*?[“”]([^“”]+)[“”]|[“”]([^“”]+)[“”][^“”\n]*?Harry'
matches = re.findall(pattern, test_string_2)
matches
````

````python
harry_dialogues = []
for line in lines:
    matches = re.findall(pattern,line)
    if matches:
        matches = matches[0]
        harry_dialogues.append(matches[0] if matches[0] else matches[1])
````

this is a little more interesting, have 2 groups from my assumption on how dialogues are written in the books. it is either

1. Harry (any other things like word `said`,`mentioned`,etc.) "quote"
2. "quote" (any other things like `said`,`mentioned,etc.) Harry

so in my pattern I use 2 groups. which is almost the same like this

```
r'Harry[^“”\n]*?[“”]([^“”]+)[“”]|[“”]([^“”]+)[“”][^“”\n]*?Harry'
```
which starts with word `Harry` and match any non-quote, non-newline characters(and make it lazy by `?`). Then match any quote and the thing inside and ending quote

the second group is the same just we match the quote first and then the name later
