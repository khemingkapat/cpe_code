# Web Scraper

## First Step: Walk before you run

Let's scrape some data!

But first thing first, we need to import necessary libraries.

````python
from bs4 import BeautifulSoup
import requests
````

If your input is a URL string, Beautiful Soup’s `.get()` method will not fetch or download the web page for you. Beautiful Soup is purely a parsing library; it cannot make network requests.

To download a web page using a URL, you must first use a HTTP library like `requests`, and then pass that downloaded content into Beautiful Soup.

````python
url = "https://quotes.toscrape.com/"

response = requests.get(url)
````

The primary purpose of the `requests` library in Python is to send HTTP requests to web servers so you can download web pages, interact with APIs, and transfer data.

It acts as a web browser without a graphical interface, allowing Python scripts to communicate directly with the internet.

And... there you go, you got some raw materials for further processing!

````python
type(response)
````

- `response.headers`: Returns a case-insensitive dictionary containing the response headers.
- `response.url`: Returns the final URL string of the response as a string.
- `response.text`: Returns the response body as a string.
- And more...

````python
response.headers
````

````python
response.url
````

````python
response.text
````

The text is still messy, noisy, and hard to read.

There is where BeautifulSoup will come in handy. It takes the raw HTML text downloaded from the internet and converts it into a *structured, searchable Python object*.

````python
soup = BeautifulSoup(response.text, "html.parser")
````

- `BeautifulSoup(...)`: Its job is to build a "parse tree" (a structured map) of the web page which allows you to easily search for specific elements like tables, links, or text instead of manually searching through thousands of lines of raw text.
- `response.text`: Raw HTML material
- `html.parser`: This tells BeautifulSoup which background engine to use to read the HTML.

````python
type(soup)
````

````python
all_tags = soup.find_all()
all_tags
````

````python
quotes = soup.find_all("span", class_="text")
````

````python
len(quotes)
````

````python
for quote in quotes[:]:
    print(quote.get_text())
````

````python
soup.find_all('table')
````

Let's first get just the title of the table.

*Note: `<th>` tag defines a header cell in an HTML table*

````python
titles = soup.find_all('th')
titles
````

Since we do not need the tags, let's clean up the data.

````python
titles_list = [title.text for title in titles]
len(titles_list)
````

````python
titles_list
````

If the output still contains newline and other symbols that are not needed, you can further clean the data using, for example, ```.strip()```
